```python
import numpy as np 
import pandas as pd 
import matplotlib.pyplot as plt
import segyio
```

```python
def load_seismic_data(filepath):
    with segyio.open(filepath,mode="r",ignore_geometry=True) as f:
        n_inlines=f.ilines
        n_crosslines=f.xlines
        f.mmap()
        data = segyio.tools.cube(f)

    return data.T

filepath = "/home/ashraf/Downloads/ST8511r92.segy"  
seismic = load_seismic_data(filepath)

print("Shape:", seismic.shape)
```
