```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt
import segyio
```

```python
# Data visualisation
seismic_file="/home/ashraf/Downloads/ST8511r92.segy"   

with segyio.open(seismic_file,"r",ignore_geometry=True) as f:

    n_traces = f.tracecount
    n_samples = len(f.samples)
    n_inlines=f.ilines
    n_crossline=f.xlines

    print("Number of traces:", n_traces)
    print("Samples per trace:", n_samples)
    print("Total number of inlines:",n_inlines)
    print("Total no. of crossline:",n_crossline)

    # Create empty array
    seismic = np.zeros((n_traces, n_samples))

    # Read traces
    for i in range(n_traces):
        seismic[i] = f.trace[i]


```
