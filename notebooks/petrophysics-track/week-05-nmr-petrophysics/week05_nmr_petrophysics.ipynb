## Theory recap

Read this first, then write the equations in your own words in the cell below.

- $\phi_{NMR} = BVI + FFI$; $BVI = TCMR - CMFF$.
- **T2 cutoff:** ~33 ms in sandstones, ~92 ms in carbonates (separates bound vs free fluid).
- **Timur-Coates:** $k = \left(\dfrac{\phi}{C}\right)^4\left(\dfrac{FFI}{BVI}\right)^2$ (φ in p.u., C≈10).
- **SDR:** $k = a\,\phi^4\,T_{2LM}^2$ — needs the T2 log-mean (not in this file; discuss it).
- NMR porosity is lithology-independent but reads low in gas (hydrogen index).
