## Theory recap

Read this first, then write the equations in your own words in the cell below.

- **Archie:** $S_w = \left(\dfrac{a\,R_w}{\phi^m R_t}\right)^{1/n}$ — clean sands only.
- **Simandoux:** adds a shale conductivity term $V_{sh}/R_{sh}$ — solve the quadratic in $S_w$.
- **Indonesia (Poupon-Leveaux):** $\dfrac{1}{\sqrt{R_t}} = \left(\dfrac{V_{sh}^{(1-V_{sh}/2)}}{\sqrt{R_{sh}}} + \dfrac{\phi^{m/2}}{\sqrt{a R_w}}\right) S_w^{n/2}$ — popular in SE Asia / Indian basins.
- **Waxman-Smits / Dual-Water:** shale effect via cation exchange (Qv / bound water) — the 'physics-based' answer.
- Shale corrections **always reduce Sw** compared with Archie in shaly intervals. Know why.
