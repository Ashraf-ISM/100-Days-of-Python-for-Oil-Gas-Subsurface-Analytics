## Theory recap

Read this first, then write the equations in your own words in the cell below.

- An image is unrolled around the borehole: x = azimuth (0–360°), y = depth.
- A planar feature appears as a sinusoid: $z(\theta) = z_0 + \dfrac{D}{2}\tan(\text{dip})\cos(\theta - \text{az})$, D = hole diameter.
- Apparent dip must be corrected for well deviation to get true dip.
- **Breakouts** (180° apart) give the minimum horizontal stress direction; **drilling-induced fractures** give the maximum.
