## Theory recap

Read this first, then write the equations in your own words in the cell below.

- Typical gradients: gas ~0.08 psi/ft, oil ~0.3–0.35 psi/ft, water ~0.43–0.45 psi/ft.
- Contact = intersection of two fitted gradient lines.
- **Excess pressure:** $P - P_{hydrostatic}$ — helps identify compartments.
- **Mobility:** $k/\mu$ from drawdown; low mobility → less reliable pressure.
- Supercharging in low-mobility zones makes points read too high.
