## Theory recap

Read this first, then write the equations in your own words in the cell below.

- **Sigma mixing:** $\Sigma_{log} = (1-\phi-V_{sh})\Sigma_{ma} + V_{sh}\Sigma_{sh} + \phi S_w\Sigma_w + \phi(1-S_w)\Sigma_{hc}$.
- Solve for $S_w$; brine Sigma rises with salinity (that is why it works).
- **Time-lapse:** $\Delta S_w$ between baseline and monitor runs.
- CO₂ has a very low Sigma (similar to gas) → CO₂ plume shows as falling Sw.
