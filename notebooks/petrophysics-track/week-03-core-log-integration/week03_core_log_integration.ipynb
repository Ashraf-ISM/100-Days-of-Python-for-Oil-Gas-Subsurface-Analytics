## Theory recap

Read this first, then write the equations in your own words in the cell below.

- **Depth matching:** core gamma vs log GR, cross-correlate to find the shift (usually 0–3 m).
- **Overburden correction:** routine core porosity is measured at ambient stress → reduce by a factor (typically 0.9–0.98).
- **Klinkenberg correction:** air permeability > liquid permeability, especially at low k.
- **Poro-perm transform:** $\log_{10}k = a\,\phi + b$ — fit per rock type, not for the whole well.
