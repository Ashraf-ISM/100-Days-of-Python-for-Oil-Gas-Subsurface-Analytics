## Theory recap

Read this first, then write the equations in your own words in the cell below.

- **Gassmann:** $K_{sat} = K_{dry} + \dfrac{(1-K_{dry}/K_{min})^2}{\phi/K_{fl} + (1-\phi)/K_{min} - K_{dry}/K_{min}^2}$, $\mu_{sat}=\mu_{dry}$.
- Moduli from logs: $K = \rho(V_p^2 - \tfrac43V_s^2)$, $\mu = \rho V_s^2$.
- **Greenberg-Castagna** Vs prediction when shear is missing.
- Dynamic Young's modulus → static via an empirical correlation; UCS from DT (e.g. McNally).
