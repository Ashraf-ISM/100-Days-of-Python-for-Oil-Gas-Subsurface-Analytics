## Theory recap

Read this first, then write the equations in your own words in the cell below.

- **Winland R35:** $\log R_{35} = 0.732 + 0.588\log k - 0.864\log\phi$ (φ in %, k in mD).
- **RQI / FZI (Amaefule):** $RQI = 0.0314\sqrt{k/\phi_e}$, $\phi_z = \phi_e/(1-\phi_e)$, $FZI = RQI/\phi_z$.
- **Lucia (carbonates):** petrophysical classes 1–3 by particle size / interparticle porosity.
- Rock types from core must then be **predicted from logs** in uncored intervals.
