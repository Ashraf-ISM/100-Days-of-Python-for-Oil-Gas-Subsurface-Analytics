## Theory recap

Read this first, then write the equations in your own words in the cell below.

- Height above FWL to Pc: $P_c = h\,\Delta\rho\,g$ (convert lab Pc with $\sigma\cos\theta$ ratios).
- **Leverett J:** $J(S_w) = \dfrac{P_c}{\sigma\cos\theta}\sqrt{\dfrac{k}{\phi}}$ — one curve per rock type.
- **Lambda model:** $S_w = a\,h^{-b} + S_{wirr}$ (simple and robust).
- **Thomeer:** hyperbola with entry pressure Pd, Bv∞ and pore geometry factor G.
