## Theory recap

Read this first, then write the equations in your own words in the cell below.

- **Pickett plot:** $\log R_t = -m\log\phi + \log(aR_w) - n\log S_w$ — water line slope = $-m$, intercept at $\phi=1$ gives $aR_w$.
- **Hingle plot:** $\phi$ vs $R_t^{-1/m}$ — straight water line through matrix point.
- **SP method:** $SSP = -K\log(R_{mf}/R_w)$, $K \approx 61 + 0.133\,T(°F)$.
- **Temperature correction (Arps):** $R_2 = R_1\dfrac{T_1+6.77}{T_2+6.77}$ (°F).
