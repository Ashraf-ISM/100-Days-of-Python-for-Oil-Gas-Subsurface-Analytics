## 🧠 What is a Perceptron?

The **perceptron** is the simplest form of a neural network — a single computational unit inspired by a biological neuron.

It takes one or more **inputs**, multiplies each by a **weight**, sums them up, adds a **bias**, and passes the result through an **activation function** to produce an output.

### Mathematical Form

$$
\hat{y} = f\left(\sum_{i=1}^{n} w_i x_i + b\right)
$$

| Symbol | Meaning |
|--------|---------|
| $x_i$ | Input features |
| $w_i$ | Weights (learned parameters) |
| $b$ | Bias term |
| $f(\cdot)$ | Activation function |
| $\hat{y}$ | Predicted output |

> 💡 The perceptron is the **building block** of all modern neural networks.

---

## ⚡ Cell 2 — Activation Functions

The activation function introduces **non-linearity** into the perceptron. Different functions serve different purposes.

### Three Common Activations

| Activation | Formula | Use Case |
|------------|---------|----------|
| **None (Identity)** | $f(x) = x$ | Regression tasks |
| **Sigmoid** | $f(x) = \dfrac{1}{1 + e^{-x}}$ | Binary probability output |
| **Step** | $f(x) = \begin{cases} 1 & x \geq 0 \\ 0 & x < 0 \end{cases}$ | Hard binary classification |

```
Input x
   │
   ▼
Weighted Sum  ──►  Activation f(·)  ──►  Output ŷ
(Σ wᵢxᵢ + b)
```

> ⚠️ The **step function** is non-differentiable at 0, which is why gradient-based training doesn't work with it directly.

---

## 📏 Cell 4 — The Perceptron Learning Rule

The perceptron learns by **adjusting its weights** whenever it makes a wrong prediction.

### Weight Update Rule

$$
w_i^{\text{new}} = w_i + \eta \cdot (y - \hat{y}) \cdot x_i
$$

$$
b^{\text{new}} = b + \eta \cdot (y - \hat{y})
$$

| Symbol | Meaning |
|--------|---------|
| $\eta$ | Learning rate (step size) |
| $y$ | True target label |
| $\hat{y}$ | Predicted output |
| $x_i$ | Corresponding input |

### Intuition

- If prediction is **correct** → $(y - \hat{y}) = 0$ → **no update**
- If prediction is **too low** → weights increase
- If prediction is **too high** → weights decrease

> 🔁 This is an **online learning** algorithm — it updates weights **one sample at a time**.

---

## 📉 Cell 5 — Cost Function: Mean Squared Error (MSE)

For regression tasks, we measure prediction quality using **Mean Squared Error**:

$$
\text{MSE} = \frac{1}{n} \sum_{i=1}^{n} (\hat{y}_i - y_i)^2
$$

### Properties of MSE

| Property | Detail |
|----------|--------|
| **Range** | $[0, \infty)$ — lower is better |
| **Perfect fit** | MSE = 0 |
| **Penalizes large errors** | Squared term amplifies outliers |
| **Differentiable** | Enables gradient-based optimization |

---

## ⛰️ Cell 6 — Gradient Descent

**Gradient Descent** is an optimization algorithm that iteratively moves weights in the direction that **reduces the cost function**.

### Core Idea

$$
w \leftarrow w + \eta \cdot \nabla_w \mathcal{L}
$$

For MSE with a perceptron:

$$
\Delta w = \eta \cdot X^T \cdot (y - \hat{y})
$$

$$
\Delta b = \eta \cdot \sum (y - \hat{y})
$$

### Algorithm Steps

```
1. Initialize weights to 0 (or small random values)
2. For each iteration:
   a. Compute prediction  ŷ = f(Xw + b)
   b. Compute error       e = y - ŷ
   c. Update weights      w ← w + η · Xᵀe
   d. Update bias         b ← b + η · Σe
   e. Check stopping criterion
3. Return final weights
```

---

## 🛑 Cell 7 — Stopping Criteria & Hyperparameters

Training doesn't run forever — we define conditions to stop.

### Stopping Conditions

| Condition | Description |
|-----------|-------------|
| **Max iterations** | Stop after a fixed number of steps |
| **Convergence threshold** | Stop when cost change $ \|C_t - C_{t-1}\| \leq \epsilon $ |

### Key Hyperparameters

| Hyperparameter | Effect |
|----------------|--------|
| **Learning Rate $\eta$** | Too large → overshooting; too small → slow convergence |
| **Iterations** | More iterations = finer optimization (up to a point) |
| **Stopping Criterion $\epsilon$** | Controls precision of convergence |
| **Initial Weights** | Can affect convergence speed and final solution |

> 🎛️ These are not learned from data — they must be **manually tuned** or searched over.

---

## 📊 Cell 8 — Perceptron vs. OLS (Ordinary Least Squares)

When fitting a line to data, how does our trained perceptron compare to classical linear regression?

### Comparison

| Aspect | Perceptron (Gradient Descent) | OLS (Analytical) |
|--------|-------------------------------|------------------|
| **Method** | Iterative optimization | Closed-form solution |
| **Exact solution?** | Approximate (depends on $\eta$, iterations) | Exact |
| **Scalable?** | ✅ Works for large datasets | ❌ Matrix inversion is expensive |
| **Flexible?** | ✅ Extendable to deep networks | ❌ Linear only |

> 💡 OLS minimizes MSE **exactly** with the formula $w = (X^T X)^{-1} X^T y$, while gradient descent **approximates** it iteratively.
